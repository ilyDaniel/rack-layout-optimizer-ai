# Rack Layout Optimizer AI

**Project Overview**

This project explores the use of **constraint programming and AI optimization techniques** to design efficient data center rack layouts.

Data centers contain racks filled with various hardware modules such as GPU nodes, CPU servers, network switches, cooling units, and storage nodes.  
The arrangement of hardware within a rack affects:

- Compute performance
- Power consumption
- Thermal efficiency
- Network performance

Manually designing optimal layouts is complex due to the number of possible configurations and infrastructure constraints.  

The goal of this project is to **automatically generate optimized rack layouts** that respect constraints while maximizing overall system performance.

# Motivation

Modern data centers must balance competing priorities:

1. **Compute Efficiency:** High-performance tasks need clustered GPU nodes.
2. **Power Management:** Racks have maximum power limits that cannot be exceeded.
3. **Thermal Management:** Overheating reduces hardware lifespan; cooling units must offset heat production.
4. **Network Efficiency:** Network switches need proximity to compute nodes for low-latency performance.
5. **Inventory Limitations:** Only a limited number of modules are available for placement.

This project simulates these real-world constraints and seeks an **AI-based solution** to efficiently place hardware modules in a rack grid.

# System Model

The system is composed of the following components:

1. **Rack**  
   - Represented as a 2D grid (width × height).  
   - Each slot can hold one module.

2. **Modules**  
   - Hardware components with properties:
     - **GPU Node:** High compute, high power, high heat
     - **CPU Node:** Medium compute, medium power, medium heat
     - **Network Switch:** Low power, high network, facilitates connectivity
     - **Cooling Unit:** Reduces heat in nearby slots
     - **Storage Node:** Low compute, medium power, medium network

3. **Inventory**  
   - Limited number of each module type available.

4. **Constraints**  
   - Hard Constraints: Must be satisfied
   - Soft Constraints: Can be violated with penalties

5. **Score Function**  
   - Evaluates the “goodness” of a rack layout based on:
     - Compute
     - Network
     - Power consumption
     - Heat
     - Adjacency bonuses

# Constraints

## Hard Constraints
- Only one module per slot
- Do not exceed inventory
- Total power ≤ rack limit (e.g., 12 kW)
- Total heat ≤ cooling capacity
- Structural placement rules (e.g., top-of-rack switches in top rows)

## Soft Constraints
- GPU nodes next to switches → network bonus
- Cooling units near high-heat modules → reduce heat
- GPU clusters → compute bonus
- Distance penalties from cooling or switches
- Stacking rules to prevent thermal hotspots

# Objective Function

Evaluate a rack layout numerically using:

score = w1 * compute_score
+ w2 * network_score
- w3 * power_penalty
- w4 * heat_penalty
+ w5 * adjacency_bonus


- `w1…w5` = weights representing priority (e.g., compute vs. network)  
- Example configurations:
  - Training Rack: prioritize compute
  - Edge Rack: prioritize network efficiency
- This allows **flexible scoring based on rack purpose**

# Optimization Approaches

1. **Constraint Solver (Google OR-Tools)**
   - Guarantees valid layouts that satisfy hard constraints
2. **Genetic Algorithm / Simulated Annealing**
   - Iterative improvement, soft constraint handling
3. **Reinforcement Learning (PyTorch)**
   - Train an agent to place modules to maximize score
   - Useful for large-scale or repeated simulations
